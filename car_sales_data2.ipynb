```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv(r"A:\practice_files\car_sales_data2.csv")
```

```python
df.duplicated().sum()
```

```python
df.info()
```

```python
df = df.drop_duplicates()
```

```python
x = df.drop("Price", axis=1)
y = df["Price"]
```

```python
x = pd.get_dummies(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
lr = LinearRegression().fit(x_train, y_train)
pred1 = lr.predict(x_test)
knn = KNeighborsRegressor().fit(x_train, y_train)
pred2 = knn.predict(x_test)
svm = SVR().fit(x_train, y_train)
pred3 = svm.predict(x_test)
dt = DecisionTreeRegressor().fit(x_train, y_train)
pred4 = dt.predict(x_test)
rf = RandomForestRegressor().fit(x_train, y_train)
pred5 = rf.predict(x_test)
```

```python
print(f"LinearRegression is: %{r2_score(y_test, pred1) * 100}")
print(f"KNeighborsRegressor is: %{r2_score(y_test, pred2) * 100}")
print(f"SVR is: %{r2_score(y_test, pred3) * 100}")
print(f"DecisionTreeRegressor is: %{r2_score(y_test, pred4) * 100}")
print(f"RandomForestRegressor is: %{r2_score(y_test, pred5) * 100}")
```
